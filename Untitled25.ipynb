```python
# ============================================================
# 0. 기본 설정
# ============================================================

from pathlib import Path
import random
import numpy as np


# ------------------------------------------------------------
# GitHub 프로젝트 폴더 기준 경로 설정
#
# 폴더 구조:
#
# move-ai_/
# ├─ data/
# │  ├─ seoul.shp
# │  ├─ seoul.dbf
# │  ├─ seoul.shx
# │  ├─ seoul.prj
# │  └─ seoul.cpg
# ├─ main.ipynb
# └─ logi_copilot_output/
# ------------------------------------------------------------

BASE_DIR = Path.cwd()

# 서울 도로망 SHP
ROAD_SHP_PATH = BASE_DIR / "data" / "seoul.shp"

# 결과 CSV 저장 폴더
OUTPUT_DIR = BASE_DIR / "logi_copilot_output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_SEED = 42

# MVP 규모
N_DRIVERS = 5
N_DELIVERIES = 20

# SHP에 속도 정보가 없을 경우 사용할 평균속도
DEFAULT_SPEED_KMH = 30.0

# OR-Tools 계산 제한시간
SOLVE_TIME_SEC = 10

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)


# ------------------------------------------------------------
# 경로 확인
# ------------------------------------------------------------

print("=" * 70)
print("LOGI-COPILOT MVP")
print("=" * 70)

print("프로젝트 폴더 :", BASE_DIR)
print("도로 SHP      :", ROAD_SHP_PATH)
print("결과 폴더     :", OUTPUT_DIR)

print("\n경로 확인")
print("프로젝트 폴더 존재 :", BASE_DIR.exists())
print("도로 SHP 존재      :", ROAD_SHP_PATH.exists())


if not ROAD_SHP_PATH.exists():
    raise FileNotFoundError(
        "\n도로 SHP 파일을 찾을 수 없습니다.\n"
        f"현재 작업 폴더: {BASE_DIR}\n"
        f"찾으려는 파일: {ROAD_SHP_PATH}\n\n"
        "GitHub 프로젝트 폴더의 data 폴더 안에 "
        "seoul.shp, seoul.dbf, seoul.shx, seoul.prj 파일이 "
        "있는지 확인해주세요."
    )
```
